# Differences Among Three Simulation Paradigms

In behavior genetics, there are typically three different types ways to simulate and test a Structural Equation Model with Family Designs. We will demonstrate the differences and performance of those three approaches with a power analysis on the vertical transmission path (f) of a Univariate SEM-PGS model. We will fit the model with OpenMx package. 
1. **Exact Cov Sample:** Derive the expected covariance matrix of all observed variables based on the SEM path diagram, and generate **one sample** with sample size $n$ directly using the covariance matrix. This means the sample follow an exact multivariate normal distribution where the mu and sigma are equal to the specified ones. 
2. **Random Cov Sample:** Derive the expected covariance matrix of all observed variables based on the SEM path diagram, and draw **$N$** samples with sample size $n$ from a population whose mu and sigma are equal to the specified ones. This means that the $N$ samples will all have different sample means and variances other than mu and sigma. How accurate the sample means and variances depends on the the sample size $n$. 
3. **Forward Time Sample:** Forward time simulation refers to the process of simulating a system’s evolution over time by starting from an initial state and then iteratively moving forward through generations. At each generation, you create a new generation based on the expected data generating model you want to test and other predefined rules. Specifically, the data generating model is logically analogous to the equations used to derive a expected covariance matrix in approach 1 and 2. Running one forward-time simulation will create one sample with $n$ and typically we need to run $N$ times to get more than one samples.
